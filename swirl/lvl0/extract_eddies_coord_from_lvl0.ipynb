```python
import pandas as pd
import numpy as np
import os
```

```python
lvl0_csv_path = r"C:\Users\leroquan\Downloads\eddy_catalogues_lvl0\eddy_catalogues_lvl0\lvl0_20230301_20230308.csv"
grid_folder = r"C:\Users\leroquan\Documents\Data\mitgcm_grids\geneva\50m"

extracted_date = '2023-03-03 00:40:00'

output_file_path = r'D:\03-eddy_detection\geneva_june_2023\eddy_catalogues_lvl0\lon_lat_eddies\lon_lat_20230603_0m.csv'
```

# Load lvl0 catalogue

```python
df_lvl0 = pd.read_csv(lvl0_csv_path)
```

```python
df_lvl0.head()
```

```python
df_lvl0.date[0]
```

# Load grid

```python
lat_grid = np.load(os.path.join(grid_folder, "lat_grid.npy"))
lon_grid = np.load(os.path.join(grid_folder, "lon_grid.npy"))
```

# Filter catalogue

```python
df_filtered = df_lvl0[(df_lvl0.depth_index == 0) & (df_lvl0.date == extracted_date)]
```

```python
df_filtered['i_eddy_cells'].iloc[1]
```

```python
np.fromstring(df_filtered['i_eddy_cells'].iloc[1].strip("[]"), sep=',')
```

```python
i_eddies = []
j_eddies = []
for _, row in df_filtered.iterrows():
    i = np.fromstring(row['i_eddy_cells'].strip("[]"), sep=',')
    j = np.fromstring(row['j_eddy_cells'].strip("[]"), sep=',')

    i_eddies.append(i)
    j_eddies.append(j)
```

```python
i_eddies = np.concat(i_eddies)
j_eddies = np.concat(j_eddies)
```

```python
i_eddies
```

```python
lat_eddies = []
lon_eddies = []
for idx in range(len(i_eddies)):
    lat = lat_grid[int(j_eddies[idx]),int(i_eddies[idx])]
    lon = lon_grid[int(j_eddies[idx]),int(i_eddies[idx])]
    lat_eddies.append(lat)
    lon_eddies.append(lon)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(lon_eddies, lat_eddies)
```

```python
coords = np.column_stack((lon_eddies, lat_eddies))
```

```python
np.savetxt(output_file_path, coords, delimiter=",", fmt="%.5f", header="longitude,latitude", comments="")
```
