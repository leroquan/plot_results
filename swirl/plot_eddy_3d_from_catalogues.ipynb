```python
# Simulated grid-like data (change with your real grid)
nx, ny = 10, 10
n_times = 5

x = np.linspace(0, 10, nx)
y = np.linspace(0, 10, ny)
xv, yv = np.meshgrid(x, y, indexing='ij')

# Flattened grid coordinates for DataFrame
df_list = []
for t in range(n_times):
    z = np.sin(xv/2 + t/2) + np.cos(yv/2 - t/3)  # just some varying depth
    df_time = pd.DataFrame({
        'time': t,
        'x': xv.ravel(),
        'y': yv.ravel(),
        'depth': z.ravel()
    })
    df_list.append(df_time)

df = pd.concat(df_list, ignore_index=True)

# Setup PyVista plotter
plotter = pv.Plotter(off_screen=True)
plot_pane = pn.pane.VTK(plotter.ren_win, sizing_mode='stretch_both')

def create_grid(time_index):
    subset = df[df['time'] == time_index]
    x_vals = subset['x'].values.reshape((nx, ny))
    y_vals = subset['y'].values.reshape((nx, ny))
    z_vals = subset['depth'].values.reshape((nx, ny))

    grid = pv.StructuredGrid(x_vals, y_vals, z_vals)
    return grid

def update_plot(time_index):
    plotter.clear()
    grid = create_grid(time_index)
    plotter.add_mesh(grid, show_edges=True, cmap="viridis")
    plotter.reset_camera()
    plot_pane.object = plotter.ren_win

# Create Panel slider
slider = pn.widgets.IntSlider(name="Time", start=0, end=n_times - 1, step=1)
slider.param.watch(lambda e: update_plot(e.new), 'value')

# Initial plot
update_plot(0)

# Display layout
dashboard = pn.Column(slider, plot_pane)
dashboard.show()
```
