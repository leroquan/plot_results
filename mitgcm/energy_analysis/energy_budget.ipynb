```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
total_ke = pd.read_csv(r"D:\04-Geneva_dummy\03_24h-wind_extended\energy_analysis\surface\ke_total_surface.csv")
```

```python
eddy_ke = pd.read_csv(r"D:\04-Geneva_dummy\03_24h-wind_extended\energy_analysis\surface\ke_surface_eddy.csv")
```

```python
seiche_ke = pd.read_csv(r"D:\04-Geneva_dummy\03_24h-wind_extended\energy_analysis\surface\ke_surface_seiche.csv")
```

```python
total_ke['date'] = pd.to_datetime(total_ke['date'])
eddy_ke['date'] = pd.to_datetime(eddy_ke['date'])
seiche_ke['time'] = pd.to_datetime(seiche_ke['time'])
```

```python
total_ke = total_ke.set_index('date')['kinetic_energy_[MJ]'] *2
eddy_ke = eddy_ke.set_index('date')['kinetic_energy_eddy_[MJ]']*2
seiche_ke = seiche_ke.set_index('time')['ke_mj_seiche']
```

```python
plt.figure(figsize=(10,5))
total_ke.plot(label = 'Total')
eddy_ke.plot(label = 'Eddy')
seiche_ke.plot(label = 'Seiche')
plt.title('Kinetic energy budget')
plt.ylabel('Kinetic energy [MJ]')
plt.xlabel('')
plt.xticks(rotation=10)
plt.legend()
#plt.savefig(r'D:\04-Geneva_dummy\03_24h-wind_extended\energy_analysis\surface\budget.png')
```

```python
plt.figure(figsize=(10,5))
(100*eddy_ke / total_ke).plot(label = 'Eddy')
(100*seiche_ke / total_ke).plot(label = 'Seiche')
plt.title('Fraction energy budget')
plt.ylabel('Fraction [%]')
plt.xlabel('')
plt.xticks(rotation=10)
plt.legend()
#plt.savefig(r'D:\04-Geneva_dummy\03_24h-wind_extended\energy_analysis\surface\fraction.png')
```
