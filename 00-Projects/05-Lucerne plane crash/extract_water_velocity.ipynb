# Get indices

```python
i_xc = np.argmin(np.abs(ds_to_plot['XC'].values - plot_point[0]))
i_yc = np.argmin(np.abs(ds_to_plot['YC'].values - plot_point[1]))
i_xg = np.argmin(np.abs(ds_to_plot['XG'].values - plot_point[0]))
i_yg = np.argmin(np.abs(ds_to_plot['YG'].values - plot_point[1]))
```

```python
start_time = datetime(2025, 7, 28, 9, 0, 0)
end_time = datetime(2025, 7, 28, 16, 0, 0)

time_index = ds_to_plot['time'].to_index()
start_idx = time_index.get_indexer([start_time], method='nearest')[0]
end_idx = time_index.get_indexer([end_time], method='nearest')[0]
```

# Get columns and rows names

```python
times = ds_to_plot['time'].isel(time=slice(start_idx, end_idx)).values
depths = ds_to_plot['Z'].values
```

```python
times
```

# Get water velocities

```python
uvel = ds_to_plot['UVEL'].isel(XG=i_xg, YC=i_yc, time=slice(start_idx, end_idx)).values
vvel = ds_to_plot['VVEL'].isel(XC=i_xc, YG=i_yg, time=slice(start_idx, end_idx)).values
```

# Get wind velocities

```python
path_run = Path(mitgcm_config['datapath'])
u10_path = path_run.parent / 'binary_data' / 'u10.bin'
v10_path = path_run.parent / 'binary_data' / 'v10.bin'
print(u10_path)
```

```python
with open(u10_path, 'rb') as fid:
    binary_u10 = np.fromfile(fid, dtype='>f8')

with open(v10_path, 'rb') as fid:
    binary_v10 = np.fromfile(fid, dtype='>f8')
```

```python
nt = len(ds_to_plot['time'])+1
nx = len(ds_to_plot['XG'])
ny = len(ds_to_plot['YC'])
```

```python
u10_array = np.reshape(binary_u10, (int(nt),int(ny),int(nx)))
v10_array = np.reshape(binary_v10, (int(nt),int(ny),int(nx)))
```

```python
u10_vel = u10_array[start_idx:end_idx, i_yc, i_xc]
v10_vel = v10_array[start_idx:end_idx, i_yc, i_xc]
```

# Rotate UVEL and VVEL to get eastward and northward velocities

```python
theta_rad = np.deg2rad(-mitgcm_config['rotation_in_degrees'])
print(theta_rad)
```

```python
u = uvel * np.cos(theta_rad) - vvel * np.sin(theta_rad)
v = uvel * np.sin(theta_rad) + vvel * np.cos(theta_rad)
```

```python
u10 = u10_vel * np.cos(theta_rad) - v10_vel * np.sin(theta_rad)
v10 = u10_vel * np.sin(theta_rad) + v10_vel * np.cos(theta_rad)
```

# Save files

```python
df_v = pd.DataFrame(v.T, index=depths, columns=times)
df_v.to_csv('water_northward_velocities_[m_s].csv')
```

```python
df_u10 = pd.DataFrame(u10.T, index=times)
df_u10.to_csv('wind_eastward_velocities_[m_s].csv')
```

```python
df_v10 = pd.DataFrame(v10.T, index=times)
df_v10.to_csv('wind_northward_velocities_[m_s].csv')
```

# Plot and save map

```python
def plot_map(i_time: int, i_depth: int):
    vmin = 17
    vmax = 21
    subsetting_factor_arrow = 4
    scale_arrow = 7
    
    
    plt.ioff()
    plt.close()
    fig, ax = plt.subplots(figsize=(20,7))
    
    T_plot = ds_to_plot['time'].values[i_time]
    Z_plot = ds_to_plot['Z'].values[i_depth]
    xr_plot = ds_to_plot['THETA'].sel(time=T_plot, Z=Z_plot)
    xr_plot = xr_plot.where(xr_plot != 0, np.nan)
    xr_plot.plot(cmap='jet', vmin=vmin, vmax=vmax)
    plt.axis('equal')

    subsetting_factor = subsetting_factor_arrow
    X_trimmed = ds_to_plot['XC'][::subsetting_factor]
    Y_trimmed = ds_to_plot['YC'][::subsetting_factor]
    U_trimmed = ds_to_plot['UVEL'].sel(time=T_plot, Z=Z_plot)[:,1:][::subsetting_factor,::subsetting_factor]
    V_trimmed = ds_to_plot['VVEL'].sel(time=T_plot, Z=Z_plot)[1:,:][::subsetting_factor,::subsetting_factor]
    
    U_trimmed = U_trimmed.where(U_trimmed != 0, np.nan)
    V_trimmed = V_trimmed.where(V_trimmed != 0, np.nan)
    plt.quiver(X_trimmed, Y_trimmed, U_trimmed, V_trimmed, scale=scale_arrow)

    plt.scatter(plot_point[0], plot_point[1], c='red', edgecolors='black', marker='X', s=60)
    
    #plt.xlim(37500,42500)
    #plt.ylim(19500,24500)
    
    plt.text(0.02, 0.98, f'{np.datetime_as_string(T_plot, unit="s").replace("T", " ")}', transform=plt.gca().transAxes, ha='left', va='top')
    plt.text(0.02, 0.9, f'Z={Z_plot:.2f}m', transform=plt.gca().transAxes, ha='left', va='top')
    plt.title('')
    
    # Remove axis, ticks, labels, and title
    ax.set_xticks([])
    ax.set_yticks([])
    ax.set_xlabel("")
    ax.set_ylabel("")
    ax.set_title("")
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['bottom'].set_visible(False)
    ax.spines['left'].set_visible(False)
    
    return plt.gcf()
```

```python
i_time = 335
i_depth = 0
plot_map(i_time, i_depth)
```

```python
i_depth = 0
depth_plot = ds_to_plot['Z'].values[i_depth]
for i_time in range(335,len(ds_to_plot['time'].values)):
    fig = plot_map(i_time, i_depth)
    fig.savefig(rf"./maps/{i_time}_{depth_plot:3.1f}m.png")
```
